```python

import ipywidgets as wid
from IPython.display import display, clear_output

# بارگذاری دیتاست
file_path = r"C:\Users\Bo$$\Desktop\taha codes\.vscode\movies_sample.json"

with open(file_path, "r", encoding="utf-8") as file:
    movies = json.load(file)

# سؤال اول: احساسات
mood = wid.RadioButtons(
    options=[
        ("شادم؛ یه چیزی بده از خنده منفجر شم!", "happy"),
        ("دلم گرفته؛ یه فیلم حال‌خوب‌کن می‌خوام.", "sad"),
        ("هیجان می‌خوام؛ آروم و قرار ندارم!", "surprise"),
        ("آرامش می‌خوام؛ مغزم مرخصی می‌خواد.", "neutral"),
        ("عصبانی‌ام؛ یه فیلم پرانرژی بده!", "angry")
    ],
    description="🎭 احساست:",
    style={"description_width": "initial"}
)

# سؤال دوم: همراه تماشا
company = wid.RadioButtons(
    options=[
        ("تنهایی؛ منم و خوراکی‌هام!", "تنهایی"),
        ("با خانواده؛ فقط جنگ نشه!", "خانواده"),
        ("با رفقا؛ فیلم یا مسخره‌بازی؟", "دوستان"),
        ("با عشق جان!", "عشق جان"),
        ("هرکی پایه باشه!", "همه")
    ],
    description="🍿 با کیا؟",
    style={"description_width": "initial"}
)

# سؤال سوم: زمان
time_choice = wid.RadioButtons(
    options=[
        ("وقت کمه؛ فیلم کوتاه!", "short"),
        ("یه فیلم با مدت معمولی.", "medium"),
        ("امشب وقتم آزاده؛ فیلم بلند!", "long")
    ],
    description="⏰ زمان:",
    style={"description_width": "initial"}
)

# سؤال چهارم: ترجیح امتیاز
rating_choice = wid.RadioButtons(
    options=[
        ("فقط فیلم‌های امتیازبالا!", "high"),
        ("تنوع مهم‌تره؛ سورپرایزم کن!", "any")
    ],
    description="🏆 امتیاز:",
    style={"description_width": "initial"}
)

output = wid.Output()

button = wid.Button(
    description = "کلیک کن فیلم ها اینجاست",
    button_style="success",
    icon="film"
)


def recommend_movies(selected_mood, selected_time, selected_rating):
    results = []

    for movie in movies:
        score = movie.get("rating", 0)

        # تطبیق با احساس
        if selected_mood in movie.get("moods", []):
            score += 3

        # زمان تقریبی فیلم؛ دیتاست فعلی مدت‌زمان ندارد،
        # بنابراین فعلاً از این معیار امتیاز نمی‌گیریم.

        # ترجیح امتیاز
        if selected_rating == "high":
            score += movie.get("rating", 0)

        results.append((score, movie))

    results.sort(key=lambda item: item[0], reverse=True)

    return [movie for score, movie in results[:3]]


def show_recommendations(_):
    with output:
        clear_output(wait=True)

        recommendations = recommend_movies(
            mood.value,
            time_choice.value,
            rating_choice.value
        )

        print("🍿فیلم مناسب تو:\n")

        for rank, movie in enumerate(recommendations, start=1):
            print(f"🏆 رتبه {rank}: {movie.get('title_fa', movie['title'])}")
            print(f"🎬 نام انگلیسی: {movie['title']}")
            print(f"📅 سال: {movie.get('year', 'نامشخص')}")
            print(f"🎭 ژانر: {', '.join(movie.get('genres', []))}")
            print(f"⭐ امتیاز: {movie.get('rating', 'نامشخص')}/10")
            print("-" * 35)


button.on_click(show_recommendations)

display(wid.VBox([
    wid.HTML("<h2> !cinMood  خوش اومدی</h2>"),
    wid.HTML("<p>!بیا فیلم مناسبت را پیدا کنیم</p>"),
    mood,
    company,
    time_choice,
    rating_choice,
    button,
    output
]))


```
